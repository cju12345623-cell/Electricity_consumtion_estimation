# Electricity Load Forecasting Baselines

Notebook version. Run from top to bottom.

```python
from pathlib import Path
import json

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.ensemble import HistGradientBoostingRegressor, RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error, mean_squared_error, r2_score
from sklearn.pipeline import make_pipeline

pd.set_option("display.max_columns", 100)
```

## 1. Settings

```python
PROJECT_ROOT = Path.cwd()
CSV_PATH = PROJECT_ROOT / "data" / "processed" / "power_load_dataset.csv"
TARGET = "load_mw"
TIME_COL = "timestamp"

REPORT_DIR = PROJECT_ROOT / "reports"
PLOT_DIR = REPORT_DIR / "plots"
REPORT_DIR.mkdir(exist_ok=True)
PLOT_DIR.mkdir(parents=True, exist_ok=True)

CSV_PATH
```

```python
PROJECT_ROOT, CSV_PATH, TARGET, TIME_COL
```

## 2. Load And Inspect Data

```python
df = pd.read_csv(CSV_PATH)
df[TIME_COL] = pd.to_datetime(df[TIME_COL], errors="coerce")
df[TARGET] = pd.to_numeric(df[TARGET], errors="coerce")
df = df.dropna(subset=[TIME_COL, TARGET]).sort_values(TIME_COL).reset_index(drop=True)

print("shape:", df.shape)
print("time range:", df[TIME_COL].min(), "~", df[TIME_COL].max())
display(df.head())
display(df.tail())
```

```python
missing = df.isna().sum().sort_values(ascending=False)
display(missing[missing > 0])

time_deltas = df[TIME_COL].sort_values().diff().dropna()
most_common_interval = time_deltas.mode().iloc[0] if not time_deltas.empty else None
irregular_intervals = int((time_deltas != most_common_interval).sum()) if most_common_interval is not None else 0

print("most common interval:", most_common_interval)
print("irregular intervals:", irregular_intervals)
```

## 3. Feature Engineering

```python
dataset = df.copy()

dataset["hour"] = dataset[TIME_COL].dt.hour
dataset["dayofweek"] = dataset[TIME_COL].dt.dayofweek
dataset["month"] = dataset[TIME_COL].dt.month
dataset["is_weekend"] = dataset["dayofweek"].isin([5, 6]).astype(int)

dataset["target_lag_24"] = dataset[TARGET].shift(24)
dataset["target_lag_168"] = dataset[TARGET].shift(168)
dataset["target_roll_24_mean"] = dataset[TARGET].shift(1).rolling(24).mean()

display(dataset.head())
```

```python
profile = {
    "rows": int(len(dataset)),
    "columns": list(dataset.columns),
    "target": TARGET,
    "time_column": TIME_COL,
    "missing_values": dataset.isna().sum().astype(int).to_dict(),
    "time_start": str(dataset[TIME_COL].min()),
    "time_end": str(dataset[TIME_COL].max()),
    "most_common_time_interval": str(most_common_interval),
    "irregular_time_steps": irregular_intervals,
}

(REPORT_DIR / "dataset_profile.json").write_text(json.dumps(profile, ensure_ascii=False, indent=2), encoding="utf-8")
profile
```

## 4. Train / Validation / Test Split

```python
train_end = int(len(dataset) * 0.70)
valid_end = int(len(dataset) * 0.85)

train_df = dataset.iloc[:train_end].copy()
valid_df = dataset.iloc[train_end:valid_end].copy()
test_df = dataset.iloc[valid_end:].copy()

print("train:", train_df.shape, train_df[TIME_COL].min(), "~", train_df[TIME_COL].max())
print("valid:", valid_df.shape, valid_df[TIME_COL].min(), "~", valid_df[TIME_COL].max())
print("test :", test_df.shape, test_df[TIME_COL].min(), "~", test_df[TIME_COL].max())
```

```python
def make_xy(frame):
    x = frame.drop(columns=[TARGET, TIME_COL])
    x = pd.get_dummies(x, drop_first=False)
    y = frame[TARGET]
    return x, y


x_train, y_train = make_xy(train_df)
x_valid, y_valid = make_xy(valid_df)
x_test, y_test = make_xy(test_df)

all_columns = sorted(set(x_train.columns) | set(x_valid.columns) | set(x_test.columns))
x_train = x_train.reindex(columns=all_columns, fill_value=0)
x_valid = x_valid.reindex(columns=all_columns, fill_value=0)
x_test = x_test.reindex(columns=all_columns, fill_value=0)

x_train.shape, x_valid.shape, x_test.shape
```

## 5. Train Models

```python
models = {
    "RandomForest": RandomForestRegressor(n_estimators=300, min_samples_leaf=2, n_jobs=-1, random_state=42),
    "HistGradientBoosting": HistGradientBoostingRegressor(max_iter=500, learning_rate=0.04, random_state=42),
}

try:
    from lightgbm import LGBMRegressor
    models["LightGBM"] = LGBMRegressor(n_estimators=500, learning_rate=0.04, num_leaves=31, random_state=42, verbose=-1)
except ImportError:
    print("LightGBM is not installed. Skipping LightGBM.")

try:
    from xgboost import XGBRegressor
    models["XGBoost"] = XGBRegressor(n_estimators=500, learning_rate=0.04, max_depth=6, objective="reg:squarederror", random_state=42)
except ImportError:
    print("XGBoost is not installed. Skipping XGBoost.")

list(models.keys())
```

```python
def regression_metrics(y_true, y_pred):
    return {
        "mae": float(mean_absolute_error(y_true, y_pred)),
        "rmse": float(mean_squared_error(y_true, y_pred) ** 0.5),
        "mape": float(mean_absolute_percentage_error(y_true, y_pred)),
        "r2": float(r2_score(y_true, y_pred)),
    }


results = {}
test_predictions = {}
trained_models = {}

for name, model in models.items():
    pipeline = make_pipeline(SimpleImputer(strategy="median"), model)
    pipeline.fit(x_train, y_train)
    trained_models[name] = pipeline
    valid_pred = pipeline.predict(x_valid)
    test_pred = pipeline.predict(x_test)
    results[name] = {
        "valid": regression_metrics(y_valid, valid_pred),
        "test": regression_metrics(y_test, test_pred),
    }
    test_predictions[name] = test_pred

(REPORT_DIR / "baseline_metrics.json").write_text(json.dumps(results, ensure_ascii=False, indent=2), encoding="utf-8")

metrics_df = pd.DataFrame({name: values["test"] for name, values in results.items()}).T.sort_values("mape")
display(metrics_df)
```

```python
# Save trained models for the website backend
import joblib

model_bundle = {
    "models": trained_models,
    "feature_columns": list(x_train.columns),
    "target": TARGET,
    "time_col": TIME_COL,
}

model_path = REPORT_DIR / "web_model_bundle.joblib"
joblib.dump(model_bundle, model_path)

print("Saved website model bundle:", model_path)
print("Models:", list(trained_models.keys()))
print("Feature count:", len(model_bundle["feature_columns"]))
```

## 6. Plot Predictions

```python
tail_size = min(300, len(y_test))
x_axis = np.arange(tail_size)

plt.figure(figsize=(14, 6))
plt.plot(x_axis, y_test.iloc[-tail_size:].to_numpy(), label="actual", linewidth=2)
for name, pred in test_predictions.items():
    plt.plot(x_axis, pred[-tail_size:], label=name, alpha=0.8)

plt.title("Power Load Forecast Baseline")
plt.xlabel("Test sample")
plt.ylabel(TARGET)
plt.legend()
plt.tight_layout()

plot_path = PLOT_DIR / "baseline_predictions.png"
plt.savefig(plot_path, dpi=150)
plt.show()

print("Saved profile:", REPORT_DIR / "dataset_profile.json")
print("Saved metrics:", REPORT_DIR / "baseline_metrics.json")
print("Saved plot:", plot_path)
```
